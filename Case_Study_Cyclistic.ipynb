```python
import pandas as pd

# Load the CSV file for initial verification
df = pd.read_csv('202601-divvy-tripdata.csv')

# Display the first 5 rows to inspect the structure
print("Preview of the first 5 rows:")
display(df.head())

# Verify total rows and columns
print(f"\nDataset dimensions: {df.shape[0]} lignes et {df.shape[1]} colonnes")

# Check column data types and missing values
print("\nGeneral information:")
df.info()
```

```python
# 1. Convert date columns to datetime format
df['started_at'] = pd.to_datetime(df['started_at'])
df['ended_at'] = pd.to_datetime(df['ended_at'])

# 2. Create the ride_length column (trip duration in seconds)
df['ride_length'] = (df['ended_at'] - df['started_at']).dt.total_seconds()

# 3. Create the day_of_week column (1 = Sunday, 7 = Saturday, per standard guideline)
# Note: In Python, .dt.dayofweek returns 0 for Monday and 6 for Sunday.
# Adjust to match the guideline format (1 = Sunday to 7 = Saturday):
df['day_of_week'] = (df['started_at'].dt.dayofweek + 1) % 7 + 1

# 4. Verify that the new columns have been successfully created
print("--- Preview with New Columns ---")
display(df[['ride_id', 'started_at', 'ended_at', 'ride_length', 'day_of_week']].head())
```

```python
import pandas as pd

# 1. List of all monthly CSV files (from October 2025 to October 2026)
files = [
    '202510-divvy-tripdata.csv', '202511-divvy-tripdata.csv', '202512-divvy-tripdata.csv',
    '202601-divvy-tripdata.csv', '202602-divvy-tripdata.csv', '202603-divvy-tripdata.csv',
    '202604-divvy-tripdata.csv', '202605-divvy-tripdata.csv', '202606-divvy-tripdata.csv',
    '202607-divvy-tripdata.csv', '202608-divvy-tripdata.csv', '202609-divvy-tripdata.csv',
    '202610-divvy-tripdata.csv'
]

dataframe_list = []

# 2. Loading and transformation loop for each monthly file
for f in files:
    print(f"Processing file: {f}...")
    temp_df = pd.read_csv(f)

    # Date transformation
    temp_df['started_at'] = pd.to_datetime(temp_df['started_at'])
    temp_df['ended_at'] = pd.to_datetime(temp_df['ended_at'])

    # Calculate trip duration
    temp_df['ride_length'] = (temp_df['ended_at'] - temp_df['started_at']).dt.total_seconds()

    # Calculate day of the week (1 = Sunday, 7 = Saturday)
    temp_df['day_of_week'] = (temp_df['started_at'].dt.dayofweek + 1) % 7 + 1

    # Append the processed dataframe to the list
    dataframe_list.append(temp_df)

# 3. Concatenate all dataframes into a single consolidated master table
df_all = pd.concat(dataframe_list, ignore_index=True)

# 4. Final check of the master dataset
print("\n--- CONSOLIDATION COMPLETE ---")
print(f"Total number of rows (trips): {df_all.shape[0]:,}")
print(f"Number of columns: {df_all.shape[1]}")
display(df_all.head())
```

```python
# Calculate cleaning metrics and filter out anomalous records
total_initial = len(df_all)

# 1. Identify negative or zero-duration trips
nb_invalides = len(df_all[df_all['ride_length'] <= 0])

# 2. Identify internal testing trips (e.g., maintenance stations)
nb_tests = len(df_all[df_all['start_station_name'] == 'HQ QR'])

# Create the final cleaned dataset
df_clean = df_all[(df_all['ride_length'] > 0) & (df_all['start_station_name'] != 'HQ QR')].copy()

total_final = len(df_clean)
lignes_supprimees = total_initial - total_final
pourcentage_retire = (lignes_supprimees / total_initial) * 100

# Display the professional cleaning report
print("--- CLEANING PHASE REPORT ---")
print(f"Initial row count: {total_initial:,}")
print(f"- Removed trips (duration <= 0 seconds): {nb_invalides:,}")
print(f"- Removed maintenance trips (HQ): {nb_tests:,}")
print(f"Total removed rows: {lignes_supprimees:,} ({pourcentage_retire:.2f}% of the dataset)")
print(f"Final clean rows retained for analysis: {total_final:,}")
```

```python
# Export the consolidated and cleaned dataframe to a single CSV file
df_clean.to_csv('divvy_tripdata_clean.csv', index=False)

print("Master dataset successfully exported to 'divvy_tripdata_clean.csv'!")
```

```python
import pandas as pd

# 1. Descriptive statistics of ride length (in minutes) by rider type
df_clean['ride_length_min'] = df_clean['ride_length'] / 60

stats_duration = df_clean.groupby('member_casual')['ride_length_min'].agg(
    Total_Trips='count',
    Mean_Duration_Min='mean',
    Median_Duration_Min='median',
    Max_Duration_Min='max'
).reset_index()

print("--- 1. DESCRIPTIVE STATISTICS OF RIDE DURATION ---")
display(stats_duration)

# 2. Usage analysis by day of the week (1 = Sunday, 7 = Saturday)
trips_by_day = df_clean.groupby(['day_of_week', 'member_casual'])['ride_id'].count().unstack()

print("\n--- 2. NUMBER OF TRIPS BY DAY OF THE WEEK ---")
display(trips_by_day)
```

```python
import pandas as pd

# 1. Extract year and month from the started_at column (format "YYYY-MM")
df_clean['year_month'] = df_clean['started_at'].dt.to_period('M')

# 2. Group by month and rider type, then count the total number of trips
trips_by_month = df_clean.groupby(['year_month', 'member_casual'])['ride_id'].count().unstack()

# 3. Ensure 'member' column is placed before 'casual'
trips_by_month = trips_by_month[['member', 'casual']]

# 4. Add a difference column (member minus casual)
trips_by_month['difference'] = trips_by_month['member'] - trips_by_month['casual']

# 5. Display the resulting table
print("--- NUMBER OF TRIPS BY MONTH WITH MEMBER VS CASUAL DIFFERENCE ---")
display(trips_by_month)
```

```python
import pandas as pd

# 1. Filter df_clean to exclude NULL values in start_station_name and end_station_name
df_no_null = df_clean[
    df_clean['start_station_name'].notna() & (df_clean['start_station_name'].str.strip() != '') &
    df_clean['end_station_name'].notna() & (df_clean['end_station_name'].str.strip() != '')
].copy()

print(f"Rows before removing NULLs: {len(df_clean):,}")
print(f"Rows after removing NULLs : {len(df_no_null):,}\n")

# --- 1. OVERALL VOLUMES & SHARES (excluding NULLs) ---
display(pd.DataFrame({
    'trips': df_no_null['member_casual'].value_counts(),
    'pct_global': df_no_null['member_casual'].value_counts(normalize=True) * 100
}).round(2))

# --- 2. MEAN & MEDIAN TRIP DURATIONS (in minutes, excluding NULLs) ---
if 'ride_length_min' not in df_no_null.columns:
    df_no_null['ride_length_min'] = df_no_null['ride_length'] / 60

display(df_no_null.groupby('member_casual')['ride_length_min'].agg(
    count='count',
    mean_min='mean',
    median_min='median'
).round(2))

# --- 3. BIKE TYPE & LOOP VS POINT-TO-POINT (% BY RIDER TYPE, excluding NULLs) ---
display((pd.crosstab(df_no_null['member_casual'], df_no_null['rideable_type'], normalize='index') * 100).round(2))

is_loop = df_no_null['start_station_name'] == df_no_null['end_station_name']
trip_type = is_loop.map({True: 'Loop', False: 'Point-to-Point'})
display((pd.crosstab(df_no_null['member_casual'], trip_type, normalize='index') * 100).round(2))

# --- 4. ZONES (Station_Zone, excluding NULLs) ---
tourist_kw = [
    "navy pier", "millennium park", "buckingham fountain", "adler planetarium",
    "field museum", "shedd aquarium", "griffin museum of science and industry",
    "dusable museum", "lincoln park zoo", "lincoln park conservatory",
    "garfield park conservatory", "theater on the lake", "mccormick place",
    "the salt shed", "obama presidential center", "beach", "harbor",
    "lakefront", "lake shore dr", "lakeshore dr", "burnham"
]

commute_kw = [
    "university", "chicago state university", "northwestern main campus",
    "sheridan rd & noyes st (nu)", "malcolm x college", "olive harvey college",
    "richard daley college", "wilbur wright college", "eths main campus",
    "wacker dr", "lasalle st", "lasalle dr", "dearborn st",
    "michigan ave & wacker dr", "michigan ave & madison st",
    "michigan ave & washington st", "michigan ave & jackson blvd",
    "canal st", "clinton st", "franklin st", "wells st",
    "orleans st & merchandise mart plaza", "fulton market",
    "daley center plaza", "city hall", "cta", "metra", "line"
]

def map_zone(val):
    s = str(val).lower()
    if any(k in s for k in tourist_kw):
        return "Tourist & Lakefront"
    if any(k in s for k in commute_kw):
        return "Commute & Campus"
    return "Other Stations"

df_no_null['Station_Zone'] = df_no_null['start_station_name'].apply(map_zone)

print("\n--- Volumes (member_casual x Station_Zone, excluding NULLs) ---")
display(pd.crosstab(df_no_null['member_casual'], df_no_null['Station_Zone'], margins=True))

print("\n--- % of total excluding NULLs (100% = entire table) ---")
display((pd.crosstab(df_no_null['member_casual'], df_no_null['Station_Zone'], normalize='all') * 100).round(2))
```
